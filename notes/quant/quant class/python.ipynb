```python
s=set(['a','c','x','d','d'])
s
```

## 3.1 Numpy 多维数组

- 索引切片  布尔索引 花式索引 
- 高效

```python
import numpy as np
#  创建一个包含股票价格的一维数组
stock_prices = np.array([100.0,105.0,98.0,102.5,99.0])
stock_prices
```

```python
# 计算股票价格平均值
average_price = np.mean(stock_prices)
print("平均价格",average_price)
```

```python
#计算标准差
import numpy as np
from decimal import getcontext, Decimal


std_dev = np.std(stock_prices)
print("标准差：", std_dev)
print(f"标准差：{std_dev:.2f}")
```

```python
import numpy as np
#创建一个二维数组 包含多只股票的价格历史数据
stock_data = np.array([[100.0,105.0,98.0,102.5,99.0],
                     [50.0,52.0,48.0,49.5,51.0],
                     [200.0,198.0,205.0,202.5,199.0]])
stock_data
```

```python
# 计算每支股票的价格
average_prices= np.mean(stock_data,axis=1)  #横着算
print("每支的股票的价格平均值",average_prices)
```

```python
average_prices_times= np.mean(stock_data,axis=0)  #数列算平均值
print("每个时间点的股票的价格平均值",average_prices_times)
```

```python
#向量化操作
#假设有一个包含股票价格的一维数组
prices = np.array([100.0,105.0,98.0,102.5,99.0])
#使用向量化操作计算每日收益率
returns = (prices[1: ] - prices [:-1]) / prices[:-1]
print("每日收益率",returns)
```

```python
prices[1:]
```

```python
prices[:-1]
```

```python
np.ones(5)/5
```

```python
#向量化计算5日移动平均线  卷积 每天赋予0.2的加权，累加
moving_average = np.convolve(prices,np.ones(5) / 5,mode = 'valid')
print("5日移动平均线",moving_average)
```

```python
#向量化计算3日移动平均线 
moving_average = np.convolve(prices,np.ones(3) / 3,mode = 'valid')
print("3日移动平均线",moving_average)
```

```python
prices[1:-1]
```

```python
prices[:-2]
```

```python
prices[2:]
```

```python
#向量化操作找到极值点 （最大值和最小值）
peak_indexes = np.where((prices[1:-1] > prices[:-2])& (prices[1:-1]>prices[2:]))[0]+1
valley_indexes = np.where((prices[1:-1] < prices[:-2])& (prices[1:-1] < prices[2:]))[0]+1

print("极大值",peak_indexes)
print("极小值",valley_indexes)
```
