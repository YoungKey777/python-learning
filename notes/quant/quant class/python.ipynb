 ## 2.2字典

- 无序 键-值对（key-value）  命名因子名称  后面跟相应的收益序列   评价指标根据收益序列计算
- 大括号 定义
- 通过键来查找和修改值 而不是位置
- 常用于表示对象属性 配置信息等

```python
stock = {
    'symbol':'AAPL',
    'name':'Apple Inc.',
    'price':150.25,
    'volume':1000000
}
```

```python
    print("股票信息")
    print("股票代码",stock['symbol'])
    print("服务名称",stock['name'])
    print("股票价格",stock['price'])
    print("成交量",stock['volume'])
```

```python
stock['price'] = 155.50
print("股票价格",stock['price'])
```

```python
#检查字典中是否存在某键
if 'close' in stock:
    print("在")
else:
    print("不在")
```

```python
#获取字典的键的列表
keys = stock.keys()
print("字典的键：",keys)
#获取字典的值的列表
values = stock.values()
print("字典的值: ",values)
```

```python
#获取所有的键值对
stock.items()
```

```python
#遍历键值对
for key,value in stock.items():
    print(key ,":",value )
```

```python
#删除某个键值对
del stock["volume"]
print(stock)
```

### 2.3集合（Set）

- 集合是无序的唯一元素集合，不允许重复
- 用大括号 或者set()函数定义 元素之间用逗号分割
- 支持集合操作 并集 交集 差集
- 常用于去重数据 检车成员资格

```python
s=set(['a','c','x','d','d'])
s
```

```python
t = set(['dd','a','cc'])
t
```

```python
s.union(t)  #并集
```

```python
s.intersection(t) #交集
```

```python
s.difference(t) #差集
```

### 2.5队列

- 运行标准差方差 优化
- 有序的数据结构
- 先进先出
- 可以使用列表或者使用queue模块中的队列类来实现
- 任务调度 广度优先搜索 缓冲

```python
from collections import deque
stack = deque()
stack.append(1)
stack.append(2)
stack
```

```python
import queue
#创建一个队列 
my_queue = queue.Queue()
```

```python
#入队操作
my_queue.put(1)
my_queue.put(2)
my_queue.put(3)
my_queue.queue
```

```python
element = my_queue.get()
element
```

```python
#查看队首元素 
my_queue.queue[0]
```

```python
is_empty = my_queue.empty()
is_empty
```
